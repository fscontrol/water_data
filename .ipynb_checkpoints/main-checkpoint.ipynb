```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import reaktoro as rkt
from scipy.optimize import curve_fit
```

```python
#evaporation coefficient
def k_evap(tair):
    return (0.575 + 0.011*tair)/580.0
def phs_caco3(tds, ca_caco3, alk_caco3, temp=30):
    a = (np.log10(tds)-1)/10
    b = -13.12*np.log10(temp+273) + 34.55
    c = np.log10(ca_caco3)-0.4
    d = np.log10(alk_caco3)
    return 9.3 + a + b - c - d
def phs_po4(ca_caco3, po4, temp=30):
    return (11.75 - np.log10(ca_caco3) - np.log10(po4) - 2*np.log10(temp))/0.65
def predict_ph(alk_caco3, a=4.17, b=1.7177):
    return a + b * np.log10(alk_caco3)
def calc_psi(row):
    ca = row.ca_caco3
    po4 = row.po4_orto
    temp = 30
    ph = row.ph
    return ph - phs_po4(ca, po4, temp)
def calc_lsi(row):
    ca = row.ca_caco3
    alk = row.alk*50
    tds = row.tds
    temp=30
    ph = row.ph
    return ph - phs_caco3(tds, ca, alk, temp)
```

```python
mu_chem = pd.read_excel("mu_water.xlsx")
mu_chem["rdate"] = pd.to_datetime(mu_chem.rdate)
mu_chem = mu_chem.set_index("rdate")
rr_chem = pd.read_excel("rr_analysis.xlsx", sheet_name="S2CT2")
mu_chem = mu_chem.loc[:,['ph', 'ec', 'tds', 'alk', 'tss', 'ca_caco3', 'cl', 'toc', 'cod', 'tvc',
       'po4_tot', 'po4_orto', 'po4_org']].resample("W").mean()
rr_chem["rdate"] = pd.to_datetime(rr_chem.rdate)
rr_chem = rr_chem.set_index("rdate").resample("W").mean()
```

```python
alk = rr_chem.alk*50 # as CaCO3
ph = rr_chem.ph
popt, pcov = curve_fit(predict_ph, alk, ph, [1.7177, 4.7177])
# стандартные ошибки параметров
perr = np.sqrt(np.diag(pcov))
print("Errors:", perr)
print(f"A={popt[0]}; B = {popt[1]}")
A = popt[0]
B = popt[1]
```

```python
rr_chem["psi"] = rr_chem.apply(calc_psi, axis=1)
rr_chem["lsi"] = rr_chem.apply(calc_lsi, axis=1)
```

```python
plt.scatter(rr_chem.tds, rr_chem.lsi)
```

```python
cycles = np.linspace(1.1, 5.0, 100)
mu_median = mu_chem.median()
mu_q25 = mu_chem.quantile(0.25)
mu_q75 = mu_chem.quantile(0.75)
```

```python
def predict_rr(mu, cycles):
    rr = mu*cycles
    rr.ph = predict_ph(rr.alk*50, A, B)
    return rr
```

```python
cycles = np.linspace(1.1, 6.0, 20)
tds = {
    "median": [],
    "q25": [],
    "q75": []
}
lsi = {
    "median": [],
    "q25": [],
    "q75": []
}
for c in cycles:
    rr_median = predict_rr(mu_median, c)
    rr_q25 = predict_rr(mu_q25, c)
    rr_q75 = predict_rr(mu_q75, c)
    tds["median"] += [rr_median.tds]
    tds["q25"] += [rr_q25.tds]
    tds["q75"] += [rr_q75.tds]
    lsi["median"] += [rr_median.ph - phs_caco3(tds=rr_median.tds, alk_caco3=rr_median.alk*50, ca_caco3=rr_median.ca_caco3)]
    lsi["q25"] += [rr_q25.ph - phs_caco3(tds=rr_q25.tds, alk_caco3=rr_q25.alk*50, ca_caco3=rr_q25.ca_caco3)]
    lsi["q75"] += [rr_q75.ph - phs_caco3(tds=rr_q75.tds, alk_caco3=rr_q75.alk*50, ca_caco3=rr_q75.ca_caco3)]
    
    
```

```python
from phreeqpython import PhreeqPython
from scipy.optimize import brentq

pp = PhreeqPython()

def build(so4_mgl):
    return pp.add_solution_raw({
        'units':       'mg/l',
        'temp':        '30',
        'pH':          '7.57',
        'Ca':          '40',
        "Mg":           "12.0",
        'Alkalinity':  '110 as CaCO3',   # 2.2 мг-экв/л
        'Cl':          '40.66',
        'S(6)':        str(so4_mgl),
        'Na':          '1.0 charge',     # <- Na добивается до нейтральности
        'P':           '0.17 as PO4',
    })

def resid(so4):
    s = build(so4)
    d = s.sc - 406.6
    s.forget()
    return d

so4 = brentq(resid, 0.1, 100.0)
sol = build(so4)
print(f'SO4 = {so4:.1f} мг/л,  Na = {sol.total("Na", units="mmol")*22.99:.1f} мг/л')
print(f'SC = {sol.sc:.0f},  SI кальцита = {sol.si("Calcite"):.2f}')
```

```python
def build_rr(cycles):
    ca = 40 * cycles
    mg = 12 * cycles
    temp=30 
    alk = 110 * cycles
    cl = 40.66 * cycles
    so4 = 10.5 * cycles
    na = 13.4 * cycles
    p = 0.17 * cycles
    return pp.add_solution_raw({
        'units':       'mg/l',
        'temp':        f'{temp}',
        'pH':          '7.57',
        'Ca':          f'{ca}',
        "Mg":           f"{mg}",
        'Alkalinity':  f'{alk} as CaCO3',   # 2.2 мг-экв/л
        'Cl':          f"{cl}",
        'S(6)':        f"{so4}",
        'Na':          f"{na}",     # <- Na добивается до нейтральности
        'P':           f'{p} as PO4',
    })
ALK_COEF = {
    'HCO3-': 1, 'CO3-2': 2, 'OH-': 1, 'H+': -1,
    'CaHCO3+': 1, 'MgHCO3+': 1, 'NaHCO3': 1,
    'CaCO3': 2, 'MgCO3': 2, 'NaCO3-': 2,
    'HPO4-2': 1, 'PO4-3': 2, 'H2PO4-': 0,
    'H3SiO4-': 1, 'B(OH)4-': 1, 'NH3': 1, 'HS-': 1,
}

def alk_meq(s):
    sp = s.species                      # моль/кг H2O
    return 1000 * sum(c * sp.get(n, 0.0) for n, c in ALK_COEF.items())
```

```python
ph = []
tds = []
lsi = []
alk = []
for c in cycles:
    p = build_rr(c)
    p.equalize(["CO2(g)"], [-3.2])
    ph += [p.pH]
    lsi += [p.si("Calcite")]
    tds += [p.sc/2.0]
    alk += [alk_meq(p)]
    
```
